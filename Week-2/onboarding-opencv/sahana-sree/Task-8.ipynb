```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("raw_captures/normal.jpeg")
```

```python
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(6,6))
plt.imshow(rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)
```

```python
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(50,50)
)
```

```python
print("Number of faces detected:", len(faces))
```

```python
output = image.copy()

for (x, y, w, h) in faces:
    cv2.rectangle(
        output,
        (x, y),
        (x+w, y+h),
        (0,255,0),
        3
    )
```

```python
plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title("Detected Face")
plt.axis("off")
plt.show()
```

```python
if len(faces) > 0:
    x, y, w, h = faces[0]

    face_crop = rgb[y:y+h, x:x+w]

    plt.figure(figsize=(5,5))
    plt.imshow(face_crop)
    plt.title("Zoomed Face Region")
    plt.axis("off")
    plt.show()
```

Observation

The Haar Cascade detector successfully detected one face in the normal-light image.A green rectangle was drawn around the face.The face was detected correctly because the image had good lighting, the face was mostly frontal, and there were no major occlusions such as sunglasses or masks.The zoomed-in face region shows clear facial features, which helped the Haar Cascade classifier detect the face accurately.
