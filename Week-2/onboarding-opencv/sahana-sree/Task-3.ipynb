```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("raw_captures/objects.jpeg")
```

```python
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
rows, cols = image.shape[:2]

M_translate = np.float32([
    [1,0,80],
    [0,1,50]
])
```

```python
translated = cv2.warpAffine(
    image,
    M_translate,
    (cols,rows)
)
```

```python
plt.imshow(cv2.cvtColor(translated,cv2.COLOR_BGR2RGB))
plt.title("Translated Image")
plt.axis("off")
plt.show()
```

```python
center = (cols//2, rows//2)

M_rotate = cv2.getRotationMatrix2D(
    center,
    36,
    1
)
```

```python
rotated = cv2.warpAffine(
    image,
    M_rotate,
    (cols,rows)
)
```

```python
plt.imshow(cv2.cvtColor(rotated,cv2.COLOR_BGR2RGB))
plt.title("Rotated Image (36°)")
plt.axis("off")
plt.show()
```

```python
scaled = cv2.resize(
    image,
    None,
    fx=1.2,
    fy=1.2
)
```

```python
plt.imshow(cv2.cvtColor(scaled,cv2.COLOR_BGR2RGB))
plt.title("Scaled Image (1.2x)")
plt.axis("off")
plt.show()
```

```python
pts1 = np.float32([
    [50,50],
    [200,50],
    [50,200]
])

pts2 = np.float32([
    [10,100],
    [200,50],
    [100,250]
])
```

```python
M_affine = cv2.getAffineTransform(
    pts1,
    pts2
)
```

```python
affine = cv2.warpAffine(
    image,
    M_affine,
    (cols,rows)
)
```

```python
plt.imshow(cv2.cvtColor(affine,cv2.COLOR_BGR2RGB))
plt.title("Affine Transformation")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(18,10))

plt.subplot(2,3,1)
plt.imshow(rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(cv2.cvtColor(translated,cv2.COLOR_BGR2RGB))
plt.title("Translated")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(cv2.cvtColor(rotated,cv2.COLOR_BGR2RGB))
plt.title("Rotated (36°)")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(cv2.cvtColor(scaled,cv2.COLOR_BGR2RGB))
plt.title("Scaled (1.2x)")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(cv2.cvtColor(affine,cv2.COLOR_BGR2RGB))
plt.title("Affine")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Reference Number: 12

Rotation Angle:
12 × 3 = 36°

Scale Factor:
1 + (2/10) = 1.2

Observation:

• Translation moved the objects to the right and downward without changing their size.

• Rotation rotated the image by 36 degrees around the center.

• Scaling enlarged the image by a factor of 1.2.

• The affine transformation changed the position and orientation of the objects while preserving straight lines.
