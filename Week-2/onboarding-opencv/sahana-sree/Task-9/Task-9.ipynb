```python
import cv2

video = cv2.VideoCapture("raw_captures/moving_object.mp4")

width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = video.get(cv2.CAP_PROP_FPS)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    "processed_video.mp4",
    fourcc,
    fps,
    (width, height),
    False
)

while True:

    ret, frame = video.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Grayscale Video", gray)

    out.write(gray)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

video.release()
out.release()
cv2.destroyAllWindows()
```

Initially, the video file could not be opened because the file path was incorrect. After correcting the path to the video, the program processed the video successfully.
