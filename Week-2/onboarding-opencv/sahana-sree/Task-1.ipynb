```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("raw_captures/lowlight.jpeg")

print(image.shape)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
```

```python
plt.figure(figsize=(16,5))

# Original (converted to RGB for correct display)
plt.subplot(1,4,1)
plt.imshow(rgb)
plt.title("Original (RGB)")
plt.axis("off")

# Grayscale
plt.subplot(1,4,2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

# RGB
plt.subplot(1,4,3)
plt.imshow(rgb)
plt.title("RGB")
plt.axis("off")

# HSV
plt.subplot(1,4,4)
plt.imshow(hsv)
plt.title("HSV")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
H = hsv[:, :, 0]
S = hsv[:, :, 1]
V = hsv[:, :, 2]
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(H, cmap="gray")
plt.title("Hue (H)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(S, cmap="gray")
plt.title("Saturation (S)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(V, cmap="gray")
plt.title("Value (Brightness)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Observation:
In my low-light image, the Value (V) channel preserves the details in the darkest parts of the image better than the Hue and Saturation channels. This is because the V channel represents the brightness of each pixel, making facial features and object boundaries more visible even under dim lighting. The Hue channel mainly stores color information, while the Saturation channel represents color intensity, both of which are less informative in dark areas where there is limited light.
