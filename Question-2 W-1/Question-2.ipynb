```python
import pandas as pd
```

```python
with open("application.log", "r") as file:
    logs = file.readlines()
```

```python
# Store ERROR and WARNING records
records = []

for line in logs:
    parts = line.strip().split()

    level = parts[2]
    module = parts[3]

    if level == "ERROR" or level == "WARNING":
        records.append({
            "Module": module,
            "Level": level
        })

# Create DataFrame
df = pd.DataFrame(records)

# Count frequency
report = df.groupby(["Module", "Level"]).size().reset_index(name="Frequency")

# Display result
print(report)

# Save CSV
report.to_csv("error_report.csv", index=False)

print("\nCSV report created successfully!")
```
